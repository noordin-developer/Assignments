```python
# 1. Student Information
name = "Ali"
age = 20
city = "Karachi"
is_student = True

print(f"Name: {name}")
print(f"Age: {age}")
print(f"City: {city}")
print(f"Student Status: {is_student}\n")
```

```python
# 2. Data Types
str_var = "Python"
int_var = 25
float_var = 89.5
bool_var = True

print(f"Value: {str_var}, Type: {type(str_var)}")
print(f"Value: {int_var}, Type: {type(int_var)}")
print(f"Value: {float_var}, Type: {type(float_var)}")
print(f"Value: {bool_var}, Type: {type(bool_var)}\n")
```

```python
# 3. Arithmetic Operations
num1 = 15
num2 = 4

print(f"Addition: {num1 + num2}")
print(f"Subtraction: {num1 - num2}")
print(f"Multiplication: {num1 * num2}")
print(f"Division: {num1 / num2}")
print(f"Modulus: {num1 % num2}")
print(f"Power: {num1 ** num2}\n")
```

```python
# 4. Student Marks Calculator
marks1 = 85
marks2 = 78
marks3 = 90

total_marks = marks1 + marks2 + marks3
average_marks = total_marks / 3

print(f"Total Marks: {total_marks}")
print(f"Average Marks: {average_marks:.2f}\n")
```

```python
# 5. Shopping Bill
price = 250.0
quantity = 3
total_bill = price * quantity

print(f"Total Bill: {total_bill}\n")
```

```python
# 6. Rectangle Area
length = 10
width = 5

area = length * width
perimeter = 2 * (length + width)

print(f"Area: {area}")
print(f"Perimeter: {perimeter}\n")
```

```python
# 7. Time Converter
minutes = 250
hours = minutes // 60
remaining_minutes = minutes % 60

print(f"Hours: {hours}")
print(f"Remaining Minutes: {remaining_minutes}\n")
```

```python
# 8. Student Eligibility
marks = 65
attendance = 80
is_eligible = (marks >= 50) and (attendance >= 75)

print(f"Is Eligible: {is_eligible}\n")
```

```python
# 9. Logical Operators
a = True
b = False

print(f"a and b: {a and b}")
print(f"a or b: {a or b}")
print(f"not a: {not a}")
print(f"not b: {not b}\n")
```

```python
# Final Challenge Student Report
student_name = "Ahmed"
student_age = 21
sub1_marks = 70
sub2_marks = 80
sub3_marks = 65
student_attendance = 82

report_total = sub1_marks + sub2_marks + sub3_marks
report_avg = report_total / 3
report_eligible = (report_avg >= 50) and (student_attendance >= 75)

print(f"Student Name: {student_name}")
print(f"Age: {student_age}")
print(f"Total Marks: {report_total}")
print(f"Average Marks: {report_avg:.2f}")
print(f"Is Eligible: {report_eligible}")
```
