### Q4

```python
# Q4 solution code here
with open("numbers.txt", "w") as file:
    for i in range(1, 6):
        file.write(str(i) + "\n")
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here
file = open("numbers.txt", "r")

total = 0

for line in file:
    total += int(line.strip())

file.close()

print("Total:", total)
```

### Q13 (Easy)
Aisi file open karne ki koshish karein jo exist nahi karti (`non_existent_file.txt`). `FileNotFoundError` ko `try-except` se handle karein.

```python
# Q13 solution code here
try:
    file = open("non_existent_file.txt", "r")
    print(file.read())
    file.close()

except FileNotFoundError:
    print("Error: File nahi mili.")
```

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
# Q17 solution code here
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print("Result:", result)

except ZeroDivisionError:
    print("Error: Number zero nahi ho sakta.")

except ValueError:
    print("Error: Please valid number enter karein.")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
# Q18 solution code here
try:
    num = int(input("Enter a number: "))
    result = 10 / num

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Number zero nahi ho sakta.")

else:
    print("Result:", result)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    else:
        print("Age is valid")


try:
    age = int(input("Enter your age: "))
    check_age(age)

except ValueError as e:
    print(e)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20 solution code here
try:
    file = open("sample.txt", "r")
    
    content = file.read()
    print(content)

except FileNotFoundError:
    print("File nahi mili.")

finally:
    try:
        file.close()
        print("File close ho gayi.")
    except:
        pass
```
